```python
# it is a branch of AI that allow computers to learn from data or patterns and make different types of prediction
```

```python
# ML means teaching to a computer using data so that it can predict pattern on new or unseen data
```

```python
# AI->ML->DL
```

```python
# we have three types of machine learning here
# 1)Supervised ML
# - the data contains both input and known output
# - WE HAVE TWO SUBTYPES OF SUPERVISED ML
# -1)Regression = used for numerical data prediction
# -2)Classification = used for categorical data prediction

# 2)Unsupervised ML
# - the data has no target / output column
# - different types of algorithm which we can use here
# -1)K-mean clustering
# -2)DBSCAN
# -3)PCA

# 3)Reinforcement ML
# the data is going to learn from action,reward or errors and penalties

# STEPS OF ML
# 1)COLLECTION OF DATA
# 2)FEATURE (INPUT VARIABLE/COLUMN)
# 3)TARGET (OUTPUT VARIABLE WHICH WE HAVE TO PREDICT)
# 4)TRAINING (TEACHING THE MODEL THROUGH DATA)
# 5)TESTING (TESTING THE MODEL ON UNSEEN DATA)
# 6)PREDICTION (MODELS OUTPUT FOR NEW DATA)
# 7)ALGORITHM (MODEL USED TO LEARN PATTERN)
# 8)ACCURACY (HOW MANY PREDICTION ARE CORRECT)
# 9)OVERFITTING (MODEL MEMORIZE TRAINING DATA TOO MUCH)
# 10)UNDERFITTING (MODEL FAILS TO LEARN IMPOERTANT PATTERN)
```
